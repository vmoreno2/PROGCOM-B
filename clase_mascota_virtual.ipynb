#TRES MÉTODOS Y TERMINAR LOS MÉTODOS QUE YA ESTÁN

```python
class Mascota():
  def __init__(self,nombre,raza,energia):
    self.nombre=nombre
    self.raza=raza
    self.hambre=50
    self.energia=energia

  def saludar(self):
    print(f"Hola,soy {self.nombre} y soy un {self.raza}")

  def comer(self):
    if self.hambre<=25:
      print("🙈")
    elif self.hambre>25 and self.hambre<=70:
      print("😋")
    elif self.hambre>70 and self.hambre<=90:
      print("😖")
    else:
      print("💀")

  def alimentar(self,c):
    comida={"leche":10,
            "polvo":50
    }
    if c in comida:
      self.hambre-=comida[c]
      self.energia+=comida[c]/2
      if self.hambre<0:
        self.hambre=0
      print(f"{self.nombre} comio {c}")
    else:
      print("eso no se come")

  def jugar(self):
    if self.energia>=5:
      self.energia-=5
      self.hambre+=10
      print(f"{self.nombre} ha jugado.\nEnergía restante:{self.energia}")
    else:
      print(f"{self.nombre} esta muy cansado para jugar")

  def dormir(self):
    self.energia+=30
    self.hambre+=10
    print(f"{self.nombre} se durmio zzz")

  def ladrar(self):
    print(f"{self.nombre}: guau guau")

  def estado(self):
    print(f"Nombre: {self.nombre}")
    print(f"Hambre: {self.hambre}")
    print(f"Energia: {self.energia}")


perro=Mascota("Firulais","chihuahua",100)
perro.saludar()
perro.comer()
perro.jugar()
perro.alimentar("leche")
perro.alimentar("pizza")
perro.ladrar()
perro.dormir()
perro.estado()
```
