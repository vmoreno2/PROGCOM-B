```python
ag = {}

while True:
    print("AGENDA")
    print("1. Ver agenda")
    print("2. Agregar actividad")
    print("3. Buscar actividad")
    print("4. Eliminar actividad")
    print("5. Mostrar cantidad de actividades")
    print("6. Ver primera actividad")
    print("7. Ver última actividad")
    print("8. Ordenar agenda")
    print("9. Salir")

    op = input("Elige una opción: ")

    if op == "1":
        if len(ag) == 0:
            print("La agenda está vacía")
        else:
            for a in ag:
                print(a, "->", ag[a])

    elif op == "2":
        n = input("Nombre de la actividad: ")
        h = input("Fecha o hora: ")
        ag[n] = h
        print("Actividad agregada")

    elif op == "3":
        n = input("Actividad a buscar: ")
        if n in ag:
            print(n, "->", ag[n])
        else:
            print("No se encontró")

    elif op == "4":
        n = input("Actividad a eliminar: ")
        if n in ag:
            del ag[n]
            print("Actividad eliminada")
        else:
            print("No se encontró")

    elif op == "5":
        print("Hay", len(ag), "actividades")

    elif op == "6":
        if len(ag) == 0:
            print("La agenda está vacía")
        else:
            p = list(ag)[0]
            print(p, "->", ag[p])

    elif op == "7":
        if len(ag) == 0:
            print("La agenda está vacía")
        else:
            u = list(ag)[-1]
            print(u, "->", ag[u])

    elif op == "8":
        ag = dict(sorted(ag.items()))
        print("Agenda ordenada")

    elif op == "9":
        print("Adiosito")
        break

    else:
        print("Opción inválida")
```
