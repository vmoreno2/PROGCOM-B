```python
flores={
    "tipo":["margarita","azucena","rosa"],
    "cant":[10,20,30],
    "precios":[1200,3000,1200]
}
compra_user = {
    'tipo': [],
    'cantidad': [],
    'precio': []
}

while True:
    print("\nQué desea hacer?")
    print(f"{1}. Comprar flores")
    print(f"{2}. Pagar")
    print(f"{3}. Generar factura")
    print(f"{4}. Salir")

    opc = input("Seleccione su opción: \n")
    match opc:
        case "1":
            print("Ha seleccionado: Comprar flores\n")

            # aca si no hay flores sale del bucle porque si no hay nada pos ya q
            if len(flores['tipo']) == 0 and len(compra_user['tipo']) == 0:
                print("Lo siento, estamos sin stock")
                break

            print(f"{1}. Agregar al carrito")
            print(f"{2}. Quitar del carrito (me arrepentí)")
            sub_opc = input("Seleccione su opción: \n")

            # si no hay stock pero hay en el carrito se pide q pague
            if sub_opc == "1":
                if len(flores['tipo']) == 0:
                    print("Ya no queda stock, puede pagar lo que tiene en el carrito")
                else:
                    # aqui se muestra el stock
                    print("Flores disponibles:")
                    for t, c, p in zip(flores['tipo'], flores['cant'], flores['precios']):
                        print(f"- {t.capitalize()}: {c} disponibles a ${p} c/u")

                    flor_opc = input("Seleccione la flor que desea comprar: ").lower().strip()
                    if flor_opc in flores['tipo']:
                        idx = flores['tipo'].index(flor_opc)
                        flor_cant = int(input("Ingrese la cantidad que desea comprar: "))

                        if flor_cant <= flores['cant'][idx]:
                            # Agregamos los valores a las listas del carrito
                            compra_user['tipo'].append(flor_opc)
                            compra_user['cantidad'].append(flor_cant)
                            compra_user['precio'].append(flores['precios'][idx])

                            # Restamos del inventario
                            flores['cant'][idx] -= flor_cant
                            print(f"¡Se han añadido {flor_cant} {flor_opc}(s) al carrito!")
                            # Removemos del inventario las flores si llegan a 0 usando .pop()
                            if flores['cant'][idx] == 0:
                                flores["tipo"].pop(idx)
                                flores["cant"].pop(idx)
                                flores["precios"].pop(idx)

                        else:
                            print("Cantidad no disponible")
                    else:
                        print(f"La flor {flor_opc} no está disponible")

            # este elif se usa para quitar del carrito en caso de arrpentirse el user
            elif sub_opc == "2":
                if len(compra_user['tipo']) == 0:
                    print("El carrito de compras está vacío. ")
                else:
                    print("En su carrito tiene:")
                    for t, c in zip(compra_user['tipo'], compra_user['cantidad']):
                        print(f"- {t.capitalize()}: {c}")

                    flor_quitar = input("Seleccione la flor que desea quitar: ").lower().strip()
                    if flor_quitar in compra_user['tipo']:
                        j = compra_user['tipo'].index(flor_quitar)
                        cant_quitar = int(input("Ingrese la cantidad que desea quitar: "))

                        if 0 < cant_quitar <= compra_user['cantidad'][j]:
                            if flor_quitar in flores['tipo']:
                                flores['cant'][flores['tipo'].index(flor_quitar)] += cant_quitar
                            else:
                                flores['tipo'].append(flor_quitar)
                                flores['cant'].append(cant_quitar)
                                flores['precios'].append(compra_user['precio'][j])

                            compra_user['cantidad'][j] -= cant_quitar
                            if compra_user['cantidad'][j] == 0:
                                compra_user['tipo'].pop(j)
                                compra_user['cantidad'].pop(j)
                                compra_user['precio'].pop(j)
                            print(f"Se han quitado {cant_quitar} {flor_quitar}(s) del carrito")
                        else:
                            print("Cantidad inválida")
                    else:
                        print(f"La flor {flor_quitar} no esta en su carrito")
            else:
                print('Numero incorrecto')
        case "2":
            if len(compra_user["tipo"]) > 0:
                #se calcula el total d lo comprado
                total_pagar = 0
                for i in range(len(compra_user["tipo"])):
                    subtotal = compra_user["cantidad"][i] * compra_user["precio"][i]
                    total_pagar += subtotal

                #tabla d comprs
                print("\nUsted compro:")
                print(f"{'Flores':<15} | {'Cantidad':<10} | {'Precio Unitario':<15} | {'Subtotal':<15}")
                print("-" * 46)
                for tipo, cantidad, precio in zip(compra_user['tipo'], compra_user['cantidad'], compra_user['precio']):
                    print(f"{tipo.capitalize():<15} | {cantidad:<10} | ${precio:<14} | ${precio*cantidad:<14}")
                print(f"\nTotal a pagar: ${total_pagar}\n")

                print("Seleccione el medio de pago: \n")
                print(f"{1}. Efectivo")
                print(f"{2}. Tarjeta de crédito")
                print(f"{3}. Tarjeta de débito")
                print(f"{4}. Otro medio de pago")
                pago_opc = input("Seleccione su opción: \n")

                print(f"\nUsted ha seleccionado la opción de pago {pago_opc}.")

                #verificacion de la money
                if pago_opc == "1" or pago_opc == "3":
                    dinero_user = int(input("¿Cuánto dinero tiene disponible? $"))
                    if dinero_user >= total_pagar:
                        print(f"Pago exitoso. Le sobran: ${dinero_user - total_pagar}")
                    else:
                        print(f"Dinero insuficiente, le faltan: ${total_pagar - dinero_user}")

                #verif a credito
                elif pago_opc == "2":
                    cuotas = int(input("¿En cuántas cuotas desea pagar?: "))
                    if cuotas > 0:
                        print(f"Pagará {cuotas} cuota(s) de ${round(total_pagar / cuotas)}")
                    else:
                        print("Número de cuotas inválido")

                else:
                    print(f"El total a pagar por su compra es: ${total_pagar}")

            else:
                print("El carrito de compras está vacío. ")

        case "3":
         print("\n" + "=" * 65)
         print("                  AVES PROGRAMADORAS")
         print("                    NIT: 9018086767")
         print("             ¡Somos evasores de impuestos!")
         print("=" * 65)
         print("                         FACTURA DE VENTA")
         print("=" * 65)

        #encabezado
         print(f"{'Flores':<15} | {'Cantidad':<10} | {'Precio Unitario':<15} | {'Subtotal':<15}")
         print("-" * 65)

        #productos comprados
         total_pagar = 0
         for tipo, cantidad, precio in zip(compra_user['tipo'], compra_user['cantidad'], compra_user['precio']):
            subtotal = cantidad * precio
            total_pagar += subtotal
            print(f"{tipo.capitalize():<15} | {cantidad:<10} | ${precio:<14} | ${subtotal:<14}")

         print("-" * 65)
         print(f"{'TOTAL A PAGAR:':<44} ${total_pagar:}")
         print("=" * 65)

        #info pago
         if len(compra_user['tipo']) > 0:
            print("                    INFORMACIÓN DEL PAGO")
            print("-" * 65)

            if pago_opc == "1":
                print(f"Medio de pago: Efectivo")
                print(f"Total recibido: ${dinero_user:}")
                print(f"Cambio: ${dinero_user - total_pagar:}")

            elif pago_opc == "2":
                print(f"Medio de pago: Tarjeta de crédito")
                print(f"Cuotas: {cuotas}")
                print(f"Valor aproximado por cuota: ${round(total_pagar / cuotas):}")
                print(f"Total recibido: Pago con tarjeta de crédito")
                print("Cambio: $0")

            elif pago_opc == "3":
                print(f"Medio de pago: Tarjeta de débito")
                print(f"Total recibido: Pago con tarjeta de débito")
                print("Cambio: $0")

            elif pago_opc == "4":
                print(f"Medio de pago: Otro medio de pago")
                print(f"Total recibido: Pago por otro medio")
                print("Cambio: $0")

            print("-" * 65)
            print("        ¡GRACIAS POR COMPRAR EN AVES PROGRAMADORAS!")
            print("=" * 65)
         else:
            print("No hay productos en la factura.")

            break
        case "4":
            print("Ha seleccionado: Salir\n")
            break
        case _:
            print('Número incorrecto, por favor seleccione uno correcto')

print("Gracias por emplear nuestros servicios. Feliz día")
```
