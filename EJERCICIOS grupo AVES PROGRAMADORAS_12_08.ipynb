```python
#9

c = input("Ingrese un carácter o escriba NO para terminar ")

if c.upper() == "NO":
    print("Programa terminado")

elif c.isdigit():
    print("Es un número")
    print("Tiene", len(c), "cifras")

elif c.lower() in "aeiou" and len(c) == 1:
    print("Es una vocal")

elif c.isalpha():

    if c.isupper():
        print("Es una letra mayúscula")
    else:
        print("Es una letra minúscula")

elif c.isalnum():
    print("Es un carácter alfanumérico")

else:
    print("Es un carácter especial")
```

```python
#10

saldo = 500000
clave = "1234"

i1 = input("Ingrese su clave: ")

if i1 == clave:
    print("Clave correcta.")

    retiro = float(input("Ingrese el valor a retirar: "))

    if retiro <= saldo:
        saldo = saldo - retiro

        print("Valor retirado:", retiro)
        print("Saldo restante:", saldo)

    else:
        print("ERROR: No tiene suficiente saldo.")

else:
    print("Clave incorrecta.")

    i2 = input("Ingrese nuevamente su clave: ")

    if i2 == clave:
        print("Clave correcta.")

        retiro = float(input("Ingrese el valor a retirar: "))

        if retiro <= saldo:
            saldo = saldo - retiro

            print("Valor retirado:", retiro)
            print("Saldo restante:", saldo)

        else:
            print("ERROR: No tiene suficiente saldo.")

    else:
        print("Clave incorrecta.")

        i3 = input("Último intento. Ingrese su clave: ")

        if i3 == clave:
            print("Clave correcta.")

            retiro = float(input("Ingrese el valor a retirar: "))

            if retiro <= saldo:
                saldo = saldo - retiro

                print("Valor retirado:", retiro)
                print("Saldo restante:", saldo)

            else:
                print("ERROR: No tiene suficiente saldo.")

        else:
            print("CLAVE INCORRECTA 3 VECES.")
            print("El cajero ha sido bloqueado.")
```

```python
#13

usuario = input("Cree su usuario: ")
password = input("Cree su contraseña: ")

tiene_numero = False
tiene_especial = False

for caracter in password:
    if caracter.isdigit():
        tiene_numero = True

    if caracter in "$/@%&":
        tiene_especial = True


if len(password) >= 10 and password[0].isupper() and tiene_numero and tiene_especial:
    print("Usuario creado correctamente")
    print("Usuario:", usuario)
    print("Contraseña válida")
else:
    print("La contraseña no cumple los requisitos")
```

```python
#14

# EJERCICIO 14 - IF

productos = {
    1: ["Arroz", 4000],
    2: ["Leche", 3500],
    3: ["Pan", 3000],
    4: ["Huevos", 18000],
    5: ["Cafe", 12000],
    6: ["Azucar", 4500],
    7: ["Aceite", 9000],
    8: ["Galletas", 2500],
    9: ["Gaseosa", 4000],
    10: ["Chocolate", 5000]
}

total = 0

print("PRODUCTOS")
for codigo, producto in productos.items():
    print(codigo, producto[0], "$", producto[1])

while True:

    codigo = int(input("Digite el código del producto (0 para terminar): "))

    if codigo == 0:
        break

    if codigo in productos:

        cantidad = int(input("Cantidad: "))

        nombre = productos[codigo][0]
        precio = productos[codigo][1]

        subtotal = precio * cantidad
        total = total + subtotal

        print(nombre, "x", cantidad, "=", subtotal)

    else:
        print("Código no válido")

iva = total * 0.19
total_final = total + iva

print("--------------------")
print("Subtotal:", total)
print("IVA:", iva)
print("TOTAL:", total_final)

dinero = float(input("Dinero recibido: "))

if dinero >= total_final:
    print("Cambio:", dinero - total_final)
else:
    print("Dinero insuficiente")

```
